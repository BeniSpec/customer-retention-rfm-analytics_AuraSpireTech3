```python
import pandas as pd

df = pd.read_csv('retention_matrix.csv')
pivot = df.pivot(index='cohort_month', columns='month_index', values='retention_rate_pct')
pivot.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
for cohort in pivot.index:
    plt.plot(pivot.columns, pivot.loc[cohort], marker='o', label=str(cohort))

plt.title('Customer Retention by Cohort')
plt.xlabel('Months Since First Purchase')
plt.ylabel('Retention Rate (%)')
plt.legend(title='Cohort', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='small')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('retention_curve.png', dpi=150)
plt.show()
```
